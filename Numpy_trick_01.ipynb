# np.sort

# Return a sorted copy of an array

# axis = 0 columns ke liye.....for ascending order ke liye

# axis = 1 rows ke liye ..... for ascending order ke liye

```python
# code
import numpy as np
a = np.random.randint(1,100,15)
a
```

```python
b = np.random.randint(1,100,24).reshape(6,4)
b
```

```python


np.sort(a)
# -----> ascending order
```

```python
np.sort(a)[::-1]
# descending order
```

```python



np.sort (b)
# rows waise sorting hua hai ....
# ascending order mein
```

```python

# agar columns waise sorting karna hai toh
#       ---> axis=0 likhana honga
np.sort(b,axis=0)
```

# numpy.concatenate() function concatenate a sequence of arrays along an existing axis .
with 2D array

```python
# code

c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)

print(c)
print(d)
```

```python
# concatenate in the form of tuple ()
# rows ke form mein
np.concatenate((c,d),axis=0)
```

```python
# concatenate in the form of tuple ()
# columns ke form mein
np.concatenate((c,d),axis=1)
```
