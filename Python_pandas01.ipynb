# Pandas Series

# A Pandas Series is like a column in a table. it is a 1D array holding data of any type

```python
import numpy as np
import pandas as pd
```

# Series from lists

```python
# string
country = ['India','USA','Japan','China','Germany']
pd.Series(country)
```

```python
# integers
runs = [13,24,56,78,100]
pd.Series(runs)
```

```python
# custom index

marks = [67,57,89,100]
subjects = ['math','science','english','hindi']

pd.Series(marks,index=subjects)
```

```python
# setting a name
marks = pd.Series(marks,index=subjects, name  ='archi ke marks')
marks
```

# Series from dict

```python
marks = {
    'math':67,
    'science':57,
    'english':89,
    'hindi':100

    }


marks_series = pd.Series(marks, name = 'archi ke marks')
marks_series
```
