# Pivot Table

* The pivot table takes simple column-wise data as input , and groups the entries into a two-dimensional table that provides a multidimensional summarization of the data

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
tips = pd.read_csv('/content/titanic.csv')
```

```python
tips
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
df.groupby('sex')[['total_bill']].mean()
```

```python
df.groupby(['sex','smoker'])[['total_bill']].mean()
```

```python
df.groupby(['sex','smoker'])[['total_bill']].mean().unstack()
```

# Pivot table

```python
df.pivot_table(values='total_bill',index='sex',columns='smoker')
```

```python
# aggfunc
df.pivot_table(values='total_bill',index='sex',columns='smoker',aggfunc='sum')
```

```python
# aggfunc
df.pivot_table(values='total_bill',index='sex',columns='smoker',aggfunc='count')
```

```python
# aggfunc
df.pivot_table(values='total_bill',index='sex',columns='smoker',aggfunc='std')
```

# all cols together

```python
 # df.pivot_table(index='sex',columns='smoker')
```

```python
# multidimensional
df
```

```python
df.pivot_table(index=['sex','smoker'],columns=['day','time'],values='total_bill')
```

```python
df.pivot_table(index=['sex','smoker'],columns=['day','time'],values='total_bill',aggfunc='mean')
```

# margins

```python
# margins
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='sum')
```

```python
# margins
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='sum',margins=True)
```
