# Set Operation

* Union(|)
* Intersection (&)
* Difference (-)
* Symmetric Difference (^)
* Membership Test
* Iteration

```python
# union ---> har ek items ek ek baar print honga
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
# s1.union(s2)
s1 | s2
```

```python
s1.union(s2)
```

```python
# intersection ----> common print honge
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1 & s2
```

```python
# Difference
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
#s1 - s2 ----> s1 ek item s2 maine present nhi hai

print(s1 - s2)
print(s2 - s1 )
```

```python
# symmetric differenceon  ----> common ko chor kr baki sab print ho jayega
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1 ^ s2
```

```python
# membership test
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
1 in s1
```

```python
1 not in s1
```

```python
# iteration
for i in s1:
    print(i)
```

# Set Functions

* len / sum / min / max / sorted
* union / updated
* intersection / intersection_update
* difference / difference_update
* symmetric_difference / symmetric_difference_updated

```python
s = {3,1,4,5,2,7}

print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s))
print(sorted(s, reverse=True))
```

```python
# union / update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.union(s2)
# s1 |s2
```

```python
# update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.update(s2)
print(s1)
print(s2)
```

```python
# intersection / intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)

s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference / difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)

s1.difference_update(s2)
print(s1)
print (s2)


```

```python
# symmetric_difference / symmetric_difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)

s1.symmetric_difference_update(s2)
print(s1)
print (s2)
```

```python
# isdisjoint / issubset / issuperset
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# ---> isdisjoint mein koi bhi number repeat nhi hona chahiye

s1.isdisjoint(s2)
```

```python
s1 = {1,2,3,4,5}
s2 = {6,7,8}

s1.isdisjoint(s2)
```

```python
# issubset
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issubset(s2)
```

```python
# issubset
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s2.issubset(s1)
```

```python
# issuperset
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issuperset(s2)
```

```python
# copy

s1 = {1,2,3}
s2 = s1.copy()

print (s1)
print(s2)
```

'gender': 'female'}

# Dictionary Comprehension


# { key : value for vars in iterable }

```python
# print 1st 10 numbers and their squares

{i:i**2 for i in range(1,11)}
```

```python
distances = {'delhi':1000, 'mumbai':2000, 'bangalore':3000}
print(distances)
```

```python
print(distances.items())
```

```python
# using existing dict

distances = {'delhi':1000, 'mumbai':2000, 'bangalore':3000}
print(distances)

{key:value*0.62 for (key,value) in distances.items()}
```

# using  Zip

```python
# using zip

days = ['sun','mon','tue','wed','thu','fri','sat']
temp = [30.5, 32.6 , 31.8, 33.4, 29.8, 30.2, 29.9]

{i:j for (i,j) in zip(days,temp)}
```

```python
# using if condition

products = {'phone':10, 'laptop':0, 'charger':50, 'tablets':0}
print(products)
```

```python
{key:value for (key , value) in products.items() if value>0}
```

```python
# nested comprehension
# print tables of number from 2 to 4
{i:{j:i*j for j in range(1,11)} for i in range(2,5)}
```
