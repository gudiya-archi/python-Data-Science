# Sort_values()
* sort_values pandas function hai jo values ko chhote se bade ya bade se chhote order mein sort karta hai

# ascending = True
* Default hota hai
*  small ---> large

# ascending = False
* large ---> small

#  sort_values() --> temporary sorting

* sort_values(inplace=True) ---> original data change

* inplace=True  ka mtlb
* "original object ke ander hi changes karo "


```python
import pandas as pd
s = pd.Series([50, 10 , 40 , 20])
print(s.sort_values())

# ye default ascending order hota hai
# (chhota ---> bade)
```

```python
print (s)
```

Descending order

```python

s.sort_values(ascending=False)
# (yaha bada ----> chhota sort hota hai )
```

# Editing Series

* using indexing
* what if an index does not exist
* slicing
* fancy indexing
* using index lable

```python
marks = {
    'math':67,
    'science':57,
    'english':89,
    'hindi':100

    }


marks_series = pd.Series(marks, name = 'archi ke marks')
marks_series
```

```python
# using indexing ----> columns ko change kr sakte hai
marks_series[1] = 100
marks_series
```

```python
# what if an index  does not exist
#  ---> yeh error nhi aayenga ---> event yeh add ho jayenga
marks_series['sst'] = 90
```

```python
marks_series
```

```python
marks_series ['python'] = 89
```

```python
marks_series
```

```python
# slicing
runs = [13,24,56,78,100]
runs_series = pd.Series(runs, name = 'runs scored by kohli')
runs_series
```

```python
runs_series[2:4] = [99,99]
runs_series
```

```python
# fancy indexing ---> edit kr diya
runs_series[[0,3,4]] = [0,0,0]
runs_series
```

```python
movies
```

```python
# using index label
movies ['2 States (2014 film)' ] = 'Alia bhatt'
movies
```
